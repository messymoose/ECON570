# 1. Note on [method chaining](https://medium.com/@ulriktpedersen/modern-pandas-streamlining-your-workflow-with-method-chaining-f65e75deb193)

Suppose, that we want to illustrate the growth of debt and surplus by country from 2000 to 2001 as follows (values are illustrative):

| Country   | Debt (2000) | Surplus (2000) | Debt (2001) | Surplus (2001) | Debt Growth | Surplus Growth |
|-----------|-------------|----------------|-------------|----------------|-------------|----------------|

However, assume that our starting point are two datasets: one for debt in 2000 and 2001 and another one for the surplus in the same years.

````python
# Creating the messy dataset part 1: df_wide_debt
df_wide_debt = (
    df.loc[df.index.get_level_values('Year').isin(['2000','2001'])] # Subset to years in 2000-2001
    .drop(columns='Surplus')                                        # Drop the surplus column
    .unstack()                                                      # Unstack the debt!
    .rename_axis(columns={'Year':None})                             # Rename the index
)
df_wide_debt.head()
````

````python
# Creating the messy dataset part 2: df_wide_surplus
df_wide_surplus = (
    df.loc[df.index.get_level_values('Year').isin(['2000','2001'])] #Subset to years in 2000-2001
    .drop(columns='Debt')                                           #Drop the debt column
    .unstack()                                                      #Unstack the surplus!
    .rename_axis(columns={'Year':None})                             #Rename the index
)
df_wide_surplus.head()
````

Now, we would like to turn this into a useful dataset. As we saw last week, we can merge these two data frames into one wide data frame using 
```python
pd.merge()
```

````python
Wide_df = pd.merge(df_wide_debt, df_wide_surplus, how = 'outer', on = 'ISO', indicator = True)
Wide_df
````

Recall that you can check how the merge went by checking the values of the `'_merge'` column created by setting `indicator = True`

````python
Wide_df['_merge'].value_counts()
````

## An Alternative: Stacking and Concatenating
We may also notice that if the two wide data frames, `df_wide_surplus` and `df_wide_debt` were in long format, then concatenating them along the columns would generate equivalent output to a merge. Let's see this in action

````python
df_long_surplus = df_wide_surplus.stack()
print(df_long_surplus.head())

#Melt alternative
df_long_surplus = (
    df_wide_surplus
    .droplevel(0, axis=1)  
    .reset_index()                      
    .melt(id_vars=["ISO"],           
          var_name="Year",          
          value_name="Surplus")  
    .set_index(['ISO','Year'])      
)

df_long_surplus.sort_index(ascending=False)
````

````python
df_long_debt =  df_wide_debt.stack()
df_long_debt

#Melt alternative
df_long_debt = (
    df_wide_debt
    .droplevel(0, axis=1)  
    .reset_index()                      
    .melt(id_vars=["ISO"],           
          var_name="Year",          
          value_name="Debt")  
    .set_index(['ISO','Year'])      
)

df_long_debt.sort_index(ascending=False)
````

````python
df_long_concat = pd.concat([df_long_debt,df_long_surplus], axis = 1) # Concatenate along axis 1
df_long_concat
````

Indeed &#x1F4A1;, we may recover `Wide_df` by unstacking this dataset

````python
Wide_df_unstacked = df_long_concat.unstack()
Wide_df_unstacked
````

**Note:** In this case, `concat` behaves like a `merge` because the two DataFrames share the same index (e.g., country or country–year). Both operations align observations based on the index labels.   However, if the indexes differ, `merge` and `concat` can yield different results depending on how the join is specified.

# Lambda Functions, .map() and Grouping &#x1F5FA;
e often need to apply operations element-wise on the columns of a DataFrame.  
The `.map()` method is a powerful way to work element-wise with a pandas Series.  

For example, suppose we want to create a new variable in `df` that measures how atypical the **surplus** or the **debt** values are. One way to do this is by computing their *z-scores*. 

````python
df_copy = df.copy()

for var in ['Debt', 'Surplus']:
    x_mean = df[var].mean()
    x_std = df[var].std()
    df_copy[var + '_zscore'] = df["Debt"].map(lambda x: (x-x_mean)/x_std)

df_copy.head()
````

How did pandas do this? Well, `.map()` can accept a python function and apply it element wise to a series. The syntax
```python
lambda x: (x-x_mean)/x_std
```
is shorthand for a formal function definition;
```python
def f(x):
    (x-x_mean)/x_std
```
Indeed, we can write
```python
f = lambda x: (x-x_mean)/x_std
```
to store the function. To see this, consider

````python
x_mean = df['Debt'].mean()
x_std = df['Debt'].std()

f = lambda x: (x-x_mean)/x_std
f(50)
````

Functions written in such a way are called *lambda functions.* The important thing to realize is that `.map()` can accept any function and a *lambda function* is just a particularly concise way of writing that function. For example we could have calculated the square of surplus in the following way.

````python
mu = df["Surplus"].mean()
sigma = df["Surplus"].std(ddof=1)

def z_map(v):
    return (v - mu) / sigma

df_copy = df.copy()
df_copy['Surplus_zscore'] = df_copy['Surplus'].map(z_map)
````

Often we like to perform operations at an aggregate level, for which we use `.groupby()`

````python
year_stats =  df['Debt'].groupby("Year").agg(['mean','std'])
year_stats.rename(columns = {'mean':'Mean_debt', 'std':'Std_debt'}, inplace=True)
year_stats.head()
````

````python

merged = df.merge(
    year_stats, 
    left_on="Year",        # use 'Year' level from MultiIndex
    right_index=True,      # match it to right's index
    how="left"
)

merged.head()
````

````python
merged['Debt_year_zscore'] = (merged['Debt']-merged['Mean_debt'])/merged['Std_debt']
merged.head()
````

# String Methods &#x1F3BB;
Pandas has a [set of methods](https://pandas.pydata.org/docs/user_guide/text.html) for working with string series element-wise. These are accessed using the syntax,
```python
MySeries.str.SomeStringMethod()
```
Let's see an example of these

````python
MessyStrings = pd.DataFrame({'Name':['aLeJanDrO HeRrEra', 'MiTCHElL vaLdes Bobes'],'Field':['Industrial Organization','Macroeconomics']})
MessyStrings
````

````python
TidyStrings = (
    MessyStrings
    .assign(Name = lambda x: x['Name'].str.title()) # Format names
)
print(TidyStrings)

# Or the alternative
print('Alternative:')
TidyStrings = MessyStrings.copy()
TidyStrings['Name'] = TidyStrings['Name'].str.title()
print(TidyStrings)
````

We can also split strings on certain characters

````python
FirstLast = TidyStrings['Name'].str.split(' ', n = 1, expand = True).rename(columns={0:'First',1:'Last'})
FirstLast
````

````python
# We can bring this new variables back into our data
pd.concat([TidyStrings,FirstLast], axis = 1)
````

There are many other useful string methods. We encourage you to [read up on them](https://pandas.pydata.org/docs/user_guide/text.html).
