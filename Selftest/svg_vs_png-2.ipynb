# Crisp Plots in Jupyter: PNG vs SVG

By default, matplotlib renders plots as PNG images which can look blurry when zoomed. SVG (vector) output stays crisp at any scale.

```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

# Mock data
x = np.linspace(0, 10, 10000)
y = sum([(0.5)**n * np.cos(4**n * x) for n in range(12)])
```

## Default PNG Output
Zoom in on this plot, you'll see it gets blurry.

```python
%config InlineBackend.figure_format = 'png'
fig, ax = plt.subplots(figsize=(6, 4))
ax.plot(x, y, 'b-', linewidth=1)
ax.set_title('PNG (default) - blurry when zoomed')
ax.set_xlabel('x')
ax.set_ylabel('y')
sns.despine()
plt.savefig("figure.jpg")
plt.show()
```

## SVG Output
We will set up the SVG in the backend:

```python
%config InlineBackend.figure_format = 'svg'
fig, ax = plt.subplots(figsize=(6, 4))
ax.plot(x, y, 'b-', linewidth=1)
ax.set_title('SVG - crisp at any zoom level')
ax.set_xlabel('x')
ax.set_ylabel('y')
sns.despine()
plt.savefig("figure.svg")
plt.show()
```

## Summary

Now go and zoom on the files, the svg is crisp!
